```python
def search_document(question, chunks, embeddings, model):
    question_embedding = model.encode([question])

    similarities = cosine_similarity(
        question_embedding,
        embeddings
    )[0]

    best_index = similarities.argmax()

    return chunks[best_index], similarities[best_index]
```

```python
vector_store_code = '''
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity


class VectorStore:

    def __init__(self):
        self.chunks = []
        self.embeddings = []

    def add(self, chunks, embeddings):
        self.chunks.extend(chunks)

        if len(self.embeddings) == 0:
            self.embeddings = np.array(embeddings)
        else:
            self.embeddings = np.vstack([
                self.embeddings,
                embeddings
            ])

    def search(self, query_embedding, top_k=3):

        if len(self.chunks) == 0:
            return []

        similarities = cosine_similarity(
            np.array(query_embedding).reshape(1, -1),
            self.embeddings
        )[0]

        top_indices = similarities.argsort()[::-1][:top_k]

        results = []

        for index in top_indices:
            results.append({
                "text": self.chunks[index],
                "score": float(similarities[index])
            })

        return results
'''

with open("services/vector_store.py", "w", encoding="utf-8") as file:
    file.write(vector_store_code)

print("vector_store.py created!")
```

```python
loader_code = '''
import os
from pypdf import PdfReader


SUPPORTED_EXTENSIONS = {
    ".txt",
    ".pdf"
}


def load_txt(file_path):
    with open(file_path, "r", encoding="utf-8") as file:
        return file.read()


def load_pdf(file_path):
    reader = PdfReader(file_path)

    text = ""

    for page in reader.pages:
        text += page.extract_text() or ""

    return text


def load_document(file_path):
    extension = os.path.splitext(file_path)[1].lower()

    if extension == ".txt":
        return load_txt(file_path)

    elif extension == ".pdf":
        return load_pdf(file_path)

    else:
        raise ValueError(
            f"Unsupported file type: {extension}"
        )
'''

with open("services/loaders.py", "w", encoding="utf-8") as file:
    file.write(loader_code)

print("loaders.py created successfully!")
```

```python
qa_engine_code = '''
from services.embedder import embed_text


def answer_question(question, vector_store, top_k=3):

    # Convert question into an embedding
    question_embedding = embed_text(question)

    # Search the vector store
    results = vector_store.search(
        question_embedding,
        top_k=top_k
    )

    if not results:
        return {
            "answer": "I could not find relevant information in the documents.",
            "sources": []
        }

    # Use the best matching result
    best_result = results[0]

    answer = best_result["text"]

    sources = results

    return {
        "answer": answer,
        "sources": sources
    }
'''

with open("services/qa_engine.py", "w", encoding="utf-8") as file:
    file.write(qa_engine_code)

print("qa_engine.py created successfully!")
```

```python
#Create Catche.py
cache_code = '''
import os
import json
import hashlib


CACHE_FOLDER = "data/cache"


def create_cache_folder():
    os.makedirs(CACHE_FOLDER, exist_ok=True)


def get_file_id(file_path):
    return hashlib.md5(
        file_path.encode("utf-8")
    ).hexdigest()


def save_cache(file_path, chunks, embeddings):
    create_cache_folder()

    file_id = get_file_id(file_path)

    cache_path = os.path.join(
        CACHE_FOLDER,
        file_id + ".json"
    )

    data = {
        "file_path": file_path,
        "chunks": chunks,
        "embeddings": embeddings.tolist()
    }

    with open(cache_path, "w", encoding="utf-8") as file:
        json.dump(data, file)

    return cache_path


def load_cache(file_path):
    create_cache_folder()

    file_id = get_file_id(file_path)

    cache_path = os.path.join(
        CACHE_FOLDER,
        file_id + ".json"
    )

    if not os.path.exists(cache_path):
        return None

    with open(cache_path, "r", encoding="utf-8") as file:
        return json.load(file)


def delete_cache(file_path):
    file_id = get_file_id(file_path)

    cache_path = os.path.join(
        CACHE_FOLDER,
        file_id + ".json"
    )

    if os.path.exists(cache_path):
        os.remove(cache_path)

        return True

    return False
'''

with open("services/cache.py", "w", encoding="utf-8") as file:
    file.write(cache_code)

print("cache.py created successfully!")
```

```python
#Creates a Conversation.py
conversation_code = '''
import os
import json
import uuid
from datetime import datetime


CONVERSATION_FOLDER = "data/conversations"


def create_conversation_folder():
    os.makedirs(CONVERSATION_FOLDER, exist_ok=True)


def create_conversation(title="New Chat"):
    create_conversation_folder()

    conversation_id = str(uuid.uuid4())

    conversation = {
        "id": conversation_id,
        "title": title,
        "created_at": datetime.now().isoformat(),
        "messages": []
    }

    file_path = os.path.join(
        CONVERSATION_FOLDER,
        conversation_id + ".json"
    )

    with open(file_path, "w", encoding="utf-8") as file:
        json.dump(conversation, file, indent=2)

    return conversation


def add_message(conversation_id, role, content):
    create_conversation_folder()

    file_path = os.path.join(
        CONVERSATION_FOLDER,
        conversation_id + ".json"
    )

    if not os.path.exists(file_path):
        return False

    with open(file_path, "r", encoding="utf-8") as file:
        conversation = json.load(file)

    conversation["messages"].append({
        "role": role,
        "content": content
    })

    with open(file_path, "w", encoding="utf-8") as file:
        json.dump(conversation, file, indent=2)

    return True


def get_conversation(conversation_id):
    file_path = os.path.join(
        CONVERSATION_FOLDER,
        conversation_id + ".json"
    )

    if not os.path.exists(file_path):
        return None

    with open(file_path, "r", encoding="utf-8") as file:
        return json.load(file)


def list_conversations():
    create_conversation_folder()

    conversations = []

    for filename in os.listdir(CONVERSATION_FOLDER):

        if filename.endswith(".json"):

            file_path = os.path.join(
                CONVERSATION_FOLDER,
                filename
            )

            with open(file_path, "r", encoding="utf-8") as file:
                conversations.append(json.load(file))

    return conversations


def delete_conversation(conversation_id):
    file_path = os.path.join(
        CONVERSATION_FOLDER,
        conversation_id + ".json"
    )

    if os.path.exists(file_path):
        os.remove(file_path)
        return True

    return False
'''

with open("services/conversations.py", "w", encoding="utf-8") as file:
    file.write(conversation_code)

print("conversations.py created successfully!")
```

```python
#Create ingest.py properly
ingest_code = '''
from services.loaders import load_document
from services.embedder import embed_texts


def create_chunks(text, chunk_size=500):

    chunks = []

    for i in range(0, len(text), chunk_size):
        chunk = text[i:i + chunk_size].strip()

        if chunk:
            chunks.append(chunk)

    return chunks


def ingest_document(file_path, vector_store):

    # Step 1: Read document
    text = load_document(file_path)

    # Step 2: Split text into chunks
    chunks = create_chunks(text)

    # Step 3: Create embeddings
    embeddings = embed_texts(chunks)

    # Step 4: Add to vector store
    vector_store.add(chunks, embeddings)

    return {
        "file_path": file_path,
        "chunks": chunks,
        "embeddings": embeddings
    }
'''

with open("services/ingest.py", "w", encoding="utf-8") as file:
    file.write(ingest_code)

print("ingest.py updated successfully!")
```

```python
#Create a simple app.py
app_code = '''
import streamlit as st

from services.vector_store import VectorStore
from services.ingest import ingest_document
from services.qa_engine import answer_question


st.set_page_config(
    page_title="DocQuery",
    page_icon="📄"
)


st.title("📄 DocQuery")
st.write("Ask questions about your documents.")


# Create VectorStore
if "vector_store" not in st.session_state:
    st.session_state.vector_store = VectorStore()


# Upload document
uploaded_file = st.file_uploader(
    "Upload a PDF or TXT file",
    type=["pdf", "txt"]
)


if uploaded_file is not None:

    file_path = "documents/" + uploaded_file.name

    with open(file_path, "wb") as file:
        file.write(uploaded_file.getbuffer())

    if st.button("Process Document"):

        ingest_document(
            file_path,
            st.session_state.vector_store
        )

        st.success("Document processed successfully!")


# Question box
question = st.chat_input(
    "Ask a question about your document..."
)


if question:

    st.chat_message("user").write(question)

    result = answer_question(
        question,
        st.session_state.vector_store
    )

    st.chat_message("assistant").write(
        result["answer"]
    )


    if result["sources"]:

        with st.expander("Sources"):

            for source in result["sources"]:

                st.write(
                    f"Score: {source['score']:.3f}"
                )

                st.write(source["text"])
'''

with open("app.py", "w", encoding="utf-8") as file:
    file.write(app_code)

print("app.py created successfully!")
```
